### Task 18

- Create a list of 5 numbers entered by the user (append them one by one
    inside a loop).
- Then print the largest number
- the smallest number
- the sum of all numbers in the list (using loops and/or list methods,
    not built-in max/min/sum).

```python
numbers = []
```

```python
num = 0
for i in range(5):
    num+=1
    inp = int(input(f"Enter #{num} number"))
    numbers.append(inp)
```

```python
numbers
```

```python
large = numbers[0]

for i in numbers:
    if i > large:
        large = i
```

```python
large
```

```python
small = numbers[0]

for i in numbers:
    if i < small:
        small = i
```

```python
small
```

```python
total = 0

for i in numbers:
    total += i
```
